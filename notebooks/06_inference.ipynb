# 06. Inferência em Novos Dados

## Contexto e Simulação de Produção
O objetivo deste notebook é simular a etapa de **inferência em ambiente de produção**. 

Em um cenário real, um modelo treinado precisa receber dados brutos inéditos (por exemplo, vindos de um formulário médico ou API), aplicar exatamente as mesmas transformações utilizadas na etapa de treinamento e retornar a previsão acompanhada da probabilidade associada.

### Fluxo de Execução:
1. Carregamento dos artefatos serializados: pré-processador (`preprocessor.joblib`) e modelo final (`best_model.joblib`).
2. Recebimento de dados brutos não processados de um novo paciente.
3. Aplicação da transformação nos dados via pipeline.
4. Geração do diagnóstico (classe prevista) e do nível de confiança (probabilidade).

```python
import pandas as pd
import joblib

# 1. Carregamento dos artefatos serializados
preprocessor = joblib.load("../models/preprocessor.joblib")
model = joblib.load("../models/best_model.joblib")

print("Artefatos carregados com sucesso:")
print("- Pipeline de Pré-processamento")
print("- Modelo Campeão (Logistic Regression)")
```

## 1. Entrada de Dados Brutos (Novo Paciente)

Criação de uma amostra sintética de um novo paciente em formato nativo (`DataFrame`), contendo as mesmas colunas do dataset original sem o `target`.

```python
# Dados brutos recebidos para inferência
new_patient_data = pd.DataFrame([{
    'age': 57,
    'sex': 1,
    'cp': 3,
    'trestbps': 145,
    'chol': 233,
    'fbs': 1,
    'restecg': 0,
    'thalach': 150,
    'exang': 0,
    'oldpeak': 2.3,
    'slope': 0,
    'ca': 0,
    'thal': 1
}])

display(new_patient_data)
```

## 2. Aplicação do Pré-processamento e Inferência

Utilizamos o método `.transform()` do `preprocessor` carregado para garantir que a imputação, o escalonamento e o One-Hot Encoding sigam rigorosamente os parâmetros aprendidos no conjunto de treino.

```python
# Transformação dos novos dados
processed_data = preprocessor.transform(new_patient_data)

# Realização da predição e cálculo da probabilidade
prediction = model.predict(processed_data)[0]
probability = model.predict_proba(processed_data)[0][1]

# Formatação do resultado
status = "Presença de Doença Cardíaca" if prediction == 1 else "Saudável / Sem Doença Cardíaca"

print("--- Resultado da Inferência ---")
print(f"Diagnóstico Preditivo : {status}")
print(f"Probabilidade Estimada : {probability * 100:.2f}%")
```
